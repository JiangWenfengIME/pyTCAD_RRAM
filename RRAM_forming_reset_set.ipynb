```python
import numpy as np
import plotly.graph_objects as go
from datetime import datetime
import os
import time
import pandas as pd
import copy
import matplotlib.pyplot as plt

from pytcad.Mesh import Mesh
from pytcad.Device import Device
from pytcad.Simulator import Simulator
from pytcad.Materials import Conductor,Oxide,Semiconductor

from scipy.interpolate import RegularGridInterpolator
from pytcad.GRF_fft import generate_correlated_field,plot_field_slices,plot_3d_isosurface,plot_volume
import dill

import pytcad.Post as post

colors = ['#1b9e77', '#d95f02', '#7570b3', '#e7298a', '#66a61e']

vtk_foldername='results_vtk/RRAM_forming_'+"{:%Y%m%d}".format(datetime.now())
if not os.path.exists(vtk_foldername):
        os.mkdir(vtk_foldername)

vtk_foldername_states='results_vtk/dill_RRAM_forming_'+"{:%Y%m%d}".format(datetime.now())
if not os.path.exists(vtk_foldername_states):
        os.mkdir(vtk_foldername_states)

start = time.time()
mesh=Mesh(mesh_type='2D',cell_type='triangular',filename='devices/RRAM/RRAM.mesh',scale=1e-9,symmetric_axis='y-axis') 
end = time.time()
print('Time mesh reading:',end-start, 's')

mesh.set_region_name(reg_num=0,reg_name='hfox')
mesh.set_region_name(reg_num=1,reg_name='BE')
mesh.set_region_name(reg_num=2,reg_name='TE')

device=Device(mesh)
device.width = 200e-9

device.set_electrode(reg_name='BE',electrode_name='BE')
device.set_electrode(reg_name='TE',electrode_name='TE')

hfox_semi=Semiconductor('hfox',mobility_model='rram_carrier_1',recombination_models=None,ion_recombination_models=['oxygen_vacancies'])
device.set_material(reg_name='hfox',type='semiconductor' ,material=hfox_semi)
device.set_material(reg_name='BE',type='conductor' ,material=Conductor(kappa=1))
device.set_material(reg_name='TE',type='conductor' ,material=Conductor(kappa=0.2))

device.set_thermal_boundary(x_min=-1,x_max=1,y_min=-0.01e-9,y_max=0.01e-9)
device.set_thermal_boundary(x_min=-1,x_max=1,y_min=29.99e-9,y_max=30.01e-9)

device.residual_ion_percentage = 0.05
device.max_residual_ion_value=1e26
device.set_uniform_ion(reg_name='hfox',conc=10,type="acceptor")

device.set_gaussian_ion(reg_name='hfox',conc=1e25,type="donor",char=2e-9,x_min=0e-9,x_max=5e-9,y_peak=20e-9)
device.set_ion_boundary_region_interface(reg_name_1='hfox',reg_name_2='TE',conc=1e25,type="donor")

conf={'method':'newton',  # 'newton' or 'gummel' method or 'ion_gummel' for nonlinear iteration
        'max_iter':100,
        'damping':0.8,
        'l2_resolution':1e-7,
        'with_electrode_voltage':False,
        'with_electron':True,
        'with_hole':False,
        'with_temperature':True,
        'with_ion':True,
        'show_details':True,
        'with_conductor_temperature':True,
        'compliance_electrode_near':'TE_1',
        'uniform_band':True,
        'compliance': False,
        'with_soret':True
        }

start = time.time()
simulator_initial=Simulator(device,conf)
end = time.time()
print('Time for setting simulator:',end-start, 's')

start = time.time()
simulator_initial.run_all()

end = time.time()
print('Time for running:',end-start, 's')

Jn,Jp=simulator_initial.get_contact_current(contact_name='TE')
print('Jn=',Jn,'A/m, Jp=',Jp,'A/m, Current=',(Jn+Jp),'A')

import pytcad.Post as post

post.export_vtk(simulator_initial, vtk_foldername, f'RRAM', x_scale=1, y_scale=1)

def iv_compliance(v):
    Ic = 1e-4
    alpha = 3.0
    i = Ic*(1-np.exp(-alpha*v))
    di_dv = Ic *(-np.exp(-alpha*v)) * (-alpha)
    return i, di_dv

vc = np.linspace(0, 2, 100)
ic, di_dv = iv_compliance(vc)
fig = go.Figure()
fig.add_trace(go.Scatter(x=vc, y=ic, mode='lines', name='I-V Curve'))
fig.add_trace(go.Scatter(x=vc, y=di_dv, mode='lines', name='dI-dV Curve'))
fig.update_layout(title='I-V Curve of Compliance Function', xaxis_title='Voltage (V)', yaxis_title='Current (A)')
fig.show()

colors = ['#1b9e77', '#d95f02', '#7570b3', '#e7298a', '#66a61e']

conf['compliance'] = True
conf['damping'] = 1.0
conf['l2_resolution'] = 1e-6
conf['with_electrode_voltage'] = True
conf['compliance_voltage_neg'] = False
conf['compliance_electrode'] = 'TE'
conf['damping_electrode_voltage'] = 1.0

conf['compliance_method'] = 'iv_func'
conf['func'] = iv_compliance

simulator=Simulator(device,conf)

for quan in simulator.quans:
        if quan['name'] == 'potential':
                for quan_1 in simulator_initial.quans:
                        if quan_1['name'] == 'potential':
                                quan['value'] = quan_1['value']
        if quan['name'] == 'electron_density':
                for quan_1 in simulator_initial.quans:
                        if quan_1['name'] == 'electron_density':
                                quan['value'] = quan_1['value']
                            

pwl_list = {"time":[0.0, 0.020, 0.04], 
        "voltage":[0.001, 2.0, 0.001]}

max_time_step = 0.0005

time_all = []
voltage = []
current = []
voltage_vte = []
simulator_trace = []

plt.figure(figsize=(6,4))
file_name_time_stepped_pvd=vtk_foldername+'/'+f'time_steped_forming'+'.pvd'
with open(file_name_time_stepped_pvd, "w") as f:
        f.write("<?xml version=\"1.0\"?>\n")
        f.write("<VTKFile type=\"Collection\" version=\"0.1\" byte_order=\"LittleEndian\" compressor=\"vtkZLibDataCompressor\">\n")
        f.write("<Collection>\n")

pvd_step = 0
stable_steps = 0          
STABLE_STEPS_REQUIRED = 5  
for i in range(len(pwl_list['time'])-1):
                    t_start = pwl_list['time'][i]
                    t_end = pwl_list['time'][i+1]
                    v_start = pwl_list['voltage'][i]
                    v_end = pwl_list['voltage'][i+1]

                    time_last = t_start
                    simulator_last = copy.deepcopy(simulator)

                    time_step = max_time_step
                    time_now = time_last + 1e-10
                    if time_now > t_end:
                            time_now = t_end
                    while time_now <= t_end:
                            vapplied = v_start + (v_end - v_start) * (time_now - t_start) / (t_end - t_start)
                            print(f'\nTime: {time_now:.4f} s, Voltage: {vapplied:.4f} V')

                            simulator.vapplied = vapplied
                            simulator.time_step = time_now - time_last

                            try:
                                    converged = simulator.run_all()    
                            except Exception as e:
                                    print(f"An error occurred during simulation at time {time_now:.4f} s: {e}")
                                    converged = False       

                            # if vapplied<0.05 or converged:
                            if converged:
                            #     post.export_vtk(simulator, vtk_foldername, f"RRAM_set_segment_{i}_{conf['compliance_electrode']}_{vapplied:.4f}", x_scale=1, y_scale=1, file_name_time_stepped_pvd=file_name_time_stepped_pvd, timestep=pvd_step)
                            
                                    pvd_step += 1
                                    Jn,Jp=simulator.get_contact_current(contact_name='TE')
                                    vte = simulator.voltage_biases['TE']
                                    print('Jn=',Jn,', Jp=',Jp,', Current=',(Jn+Jp),'\n')
                                    print(f'Voltage on TE: {vte:.5f} V\n')

                                    post.export_vtk(simulator, vtk_foldername, f"RRAM_set_segment_{i}_{conf['compliance_electrode']}_{vte:.8f}_{vapplied:.8f}", x_scale=1, y_scale=1, file_name_time_stepped_pvd=file_name_time_stepped_pvd, timestep=pvd_step)

                                    time_all.append(time_now)
                                    current.append(Jn+Jp)
                                    voltage.append(vapplied)
                                    voltage_vte.append(vte)

                                    plt.clf()
                                    plt.plot(voltage, np.abs(current), color=colors[0], marker='o', markerfacecolor='w', linestyle='None',label='RRAM set')
                                    
                                    plt.xlabel('Voltage (V)')
                                    plt.ylabel('Current (A)')
                                    plt.yscale('log')
                                    plt.title('RRAM set simulation')
                                    plt.grid(True)
                                    plt.savefig(vtk_foldername+f'/IV_Forming.png')

                                    time_last = time_now
                                    simulator_last = copy.deepcopy(simulator)
                            
                                    # 修改：稳定步数累加，达到阈值后才增大 time_step
                                    stable_steps += 1
                                    print(f'Stable steps after convergence: {stable_steps}/{STABLE_STEPS_REQUIRED}')
                                    if stable_steps >= STABLE_STEPS_REQUIRED:
                                            time_step = 1.5 * time_step
                                            if time_step > max_time_step:
                                                    time_step = max_time_step
                                            stable_steps = 0  # 重置计数器
                                            print(f'New time step after {STABLE_STEPS_REQUIRED} stable steps: {time_step:.6e} s')
                                    else:
                                            print(f'Holding time step at: {time_step:.6e} s (waiting for {STABLE_STEPS_REQUIRED} stable steps)')

                                    if time_now == t_end:
                                            break
                                    elif time_now + time_step <= t_end:
                                            time_now += time_step
                                    elif time_now + time_step > t_end:
                                            time_now = t_end
                            else:
                                    print(f'Simulation did not converge at time: {time_now:.4f} s and voltage: {vapplied:.4f} V')
                                    simulator = copy.deepcopy(simulator_last)

                                    time_step = time_step/1.5
                                    print(f'New time step after non-convergence: {time_step:.6e} s')

                                    time_now = time_last + time_step
                            

with open(file_name_time_stepped_pvd, "a") as f:
        f.write("  </Collection>\n")
        f.write("</VTKFile>\n")

simulator_forming=copy.deepcopy(simulator)

trace_forming=go.Scatter(x=voltage, y=np.abs(current), mode='markers+lines', 
                line=dict(color=colors[2]), name='RRAM_set')

trace_v_time=go.Scatter(x=time_all, y=voltage, mode='markers+lines', 
                line=dict(color=colors[0]), name='Vapplied')
trace_vte_time=go.Scatter(x=time_all, y=voltage_vte, mode='markers+lines', 
                line=dict(color=colors[1]), name='Voltage on Device')

################################################################################################
#########################################  Serilization  #######################################
################################################################################################

file_name_plot = f"simulator_forming.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(simulator_forming, f)

file_name_plot = f"trace_forming.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(trace_forming, f)

file_name_plot = f"trace_v_time.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(trace_v_time, f)

file_name_plot = f"trace_vte_time.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(trace_vte_time, f)

file_name_plot = f"voltage.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(voltage, f)

file_name_plot = f"current.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(current, f)

file_name_plot = f"time_all.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(time_all, f)

file_name_plot = f"voltage_vte.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(voltage_vte, f)
```

```python
import numpy as np
import plotly.graph_objects as go
from datetime import datetime
import os
import time
import pandas as pd
import copy
import matplotlib.pyplot as plt

from pytcad.Mesh import Mesh
from pytcad.Device import Device
from pytcad.Simulator import Simulator
from pytcad.Materials import Conductor,Oxide,Semiconductor

from scipy.interpolate import RegularGridInterpolator
from pytcad.GRF_fft import generate_correlated_field,plot_field_slices,plot_3d_isosurface,plot_volume
import dill

import pytcad.Post as post

colors = ['#1b9e77', '#d95f02', '#7570b3', '#e7298a', '#66a61e']

vtk_foldername='results_vtk/RRAM_set_'+"{:%Y%m%d}".format(datetime.now())
if not os.path.exists(vtk_foldername):
        os.mkdir(vtk_foldername)

vtk_foldername_states='results_vtk/dill_RRAM_set_'+"{:%Y%m%d}".format(datetime.now())
if not os.path.exists(vtk_foldername_states):
        os.mkdir(vtk_foldername_states)

start = time.time()
mesh=Mesh(mesh_type='2D',cell_type='triangular',filename='devices/RRAM/RRAM.mesh',scale=1e-9,symmetric_axis='y-axis') 
end = time.time()
print('Time mesh reading:',end-start, 's')

mesh.set_region_name(reg_num=0,reg_name='hfox')
mesh.set_region_name(reg_num=1,reg_name='BE')
mesh.set_region_name(reg_num=2,reg_name='TE')

device=Device(mesh)
device.width = 200e-9

device.set_electrode(reg_name='BE',electrode_name='BE')
device.set_electrode(reg_name='TE',electrode_name='TE')

hfox_semi=Semiconductor('hfox',mobility_model='rram_carrier_1',recombination_models=None,ion_recombination_models=['oxygen_vacancies'])
device.set_material(reg_name='hfox',type='semiconductor' ,material=hfox_semi)
device.set_material(reg_name='BE',type='conductor' ,material=Conductor(kappa=1))
device.set_material(reg_name='TE',type='conductor' ,material=Conductor(kappa=0.2))

device.set_thermal_boundary(x_min=-1,x_max=1,y_min=-0.01e-9,y_max=0.01e-9)
device.set_thermal_boundary(x_min=-1,x_max=1,y_min=29.99e-9,y_max=30.01e-9)

device.residual_ion_percentage = 0.05
device.max_residual_ion_value=1e26
device.set_uniform_ion(reg_name='hfox',conc=10,type="acceptor")

device.set_gaussian_ion(reg_name='hfox',conc=1e25,type="donor",char=2e-9,x_min=0e-9,x_max=5e-9,y_peak=20e-9)
device.set_ion_boundary_region_interface(reg_name_1='hfox',reg_name_2='TE',conc=1e25,type="donor")

conf={'method':'newton',  # 'newton' or 'gummel' method or 'ion_gummel' for nonlinear iteration
        'max_iter':100,
        'damping':0.8,
        'l2_resolution':1e-7,
        'with_electrode_voltage':False,
        'with_electron':True,
        'with_hole':False,
        'with_temperature':True,
        'with_ion':True,
        'show_details':True,
        'with_conductor_temperature':True,
        'compliance_electrode_near':'TE_1',
        'uniform_band':True,
        'compliance': False,
        'with_soret':True
        }

start = time.time()
simulator_initial=Simulator(device,conf)
end = time.time()
print('Time for setting simulator:',end-start, 's')

start = time.time()
simulator_initial.run_all()

end = time.time()
print('Time for running:',end-start, 's')

Jn,Jp=simulator_initial.get_contact_current(contact_name='TE')
print('Jn=',Jn,'A/m, Jp=',Jp,'A/m, Current=',(Jn+Jp),'A')

import pytcad.Post as post

post.export_vtk(simulator_initial, vtk_foldername, f'RRAM', x_scale=1, y_scale=1)

def iv_compliance(v):
    Ic = 1e-4
    alpha = 3.0
    i = Ic*(1-np.exp(-alpha*v))
    di_dv = Ic *(-np.exp(-alpha*v)) * (-alpha)
    return i, di_dv

vc = np.linspace(0, 2, 100)
ic, di_dv = iv_compliance(vc)
fig = go.Figure()
fig.add_trace(go.Scatter(x=vc, y=ic, mode='lines', name='I-V Curve'))
fig.add_trace(go.Scatter(x=vc, y=di_dv, mode='lines', name='dI-dV Curve'))
fig.update_layout(title='I-V Curve of Compliance Function', xaxis_title='Voltage (V)', yaxis_title='Current (A)')
fig.show()

colors = ['#1b9e77', '#d95f02', '#7570b3', '#e7298a', '#66a61e']

conf['compliance'] = True
conf['damping'] = 1.0
conf['l2_resolution'] = 1e-6
conf['with_electrode_voltage'] = True
conf['compliance_voltage_neg'] = False
conf['compliance_electrode'] = 'TE'
conf['damping_electrode_voltage'] = 1.0

conf['compliance_method'] = 'iv_func'
conf['func'] = iv_compliance


with open('results_vtk/dill_RRAM_reset_20260509/simulator_reset.pkl', 'rb') as f:
    simulator_reset = dill.load(f)

device.ion_density = copy.deepcopy(simulator_reset.ion_density)
device.ion_density_residual = copy.deepcopy(simulator_reset.ion_density_residual)
simulator=Simulator(device,conf)
simulator.ion_density_max =  copy.deepcopy(simulator_reset.ion_density_max)

for quan in simulator.quans:
        if quan['name'] == 'potential':
                for quan_1 in simulator_initial.quans:
                        if quan_1['name'] == 'potential':
                                quan['value'] = quan_1['value']
        if quan['name'] == 'electron_density':
                for quan_1 in simulator_initial.quans:
                        if quan_1['name'] == 'electron_density':
                                quan['value'] = quan_1['value']
                            

pwl_list = {"time":[0.0, 0.015, 0.03], 
        "voltage":[0.001, 1.5, 0.001]}
# max_time_step = 0.001
max_time_step = 0.0005

time_all = []
voltage = []
current = []
voltage_vte = []
simulator_trace = []

plt.figure(figsize=(6,4))
file_name_time_stepped_pvd=vtk_foldername+'/'+f'time_steped_set'+'.pvd'
with open(file_name_time_stepped_pvd, "w") as f:
        f.write("<?xml version=\"1.0\"?>\n")
        f.write("<VTKFile type=\"Collection\" version=\"0.1\" byte_order=\"LittleEndian\" compressor=\"vtkZLibDataCompressor\">\n")
        f.write("<Collection>\n")

pvd_step = 0
stable_steps = 0          # 新增：收敛后的稳定步数计数器
STABLE_STEPS_REQUIRED = 5  # 新增：需要稳定的步数阈值
for i in range(len(pwl_list['time'])-1):
                    t_start = pwl_list['time'][i]
                    t_end = pwl_list['time'][i+1]
                    v_start = pwl_list['voltage'][i]
                    v_end = pwl_list['voltage'][i+1]

                    time_last = t_start
                    simulator_last = copy.deepcopy(simulator)

                    time_step = max_time_step
                    time_now = time_last + 1e-10
                    if time_now > t_end:
                            time_now = t_end
                    while time_now <= t_end:
                            vapplied = v_start + (v_end - v_start) * (time_now - t_start) / (t_end - t_start)
                            print(f'\nTime: {time_now:.4f} s, Voltage: {vapplied:.4f} V')

                            simulator.vapplied = vapplied
                            simulator.time_step = time_now - time_last

                            try:
                                    converged = simulator.run_all()    
                            except Exception as e:
                                    print(f"An error occurred during simulation at time {time_now:.4f} s: {e}")
                                    converged = False       

                            # if vapplied<0.05 or converged:
                            if converged:
                            #     post.export_vtk(simulator, vtk_foldername, f"RRAM_set_segment_{i}_{conf['compliance_electrode']}_{vapplied:.4f}", x_scale=1, y_scale=1, file_name_time_stepped_pvd=file_name_time_stepped_pvd, timestep=pvd_step)
                            
                                    pvd_step += 1
                                    Jn,Jp=simulator.get_contact_current(contact_name='TE')
                                    vte = simulator.voltage_biases['TE']
                                    print('Jn=',Jn,', Jp=',Jp,', Current=',(Jn+Jp),'\n')
                                    print(f'Voltage on TE: {vte:.5f} V\n')

                                    post.export_vtk(simulator, vtk_foldername, f"RRAM_set_segment_{i}_{conf['compliance_electrode']}_{vte:.8f}_{vapplied:.8f}", x_scale=1, y_scale=1, file_name_time_stepped_pvd=file_name_time_stepped_pvd, timestep=pvd_step)

                                    time_all.append(time_now)
                                    current.append(Jn+Jp)
                                    voltage.append(vapplied)
                                    voltage_vte.append(vte)

                                    plt.clf()
                                    plt.plot(voltage, np.abs(current), color=colors[0], marker='o', markerfacecolor='w', linestyle='None',label='RRAM set')
                                    
                                    plt.xlabel('Voltage (V)')
                                    plt.ylabel('Current (A)')
                                    plt.yscale('log')
                                    plt.title('RRAM set simulation')
                                    plt.grid(True)
                                    plt.savefig(vtk_foldername+f'/IV_Set.png')

                                    time_last = time_now
                                    simulator_last = copy.deepcopy(simulator)
                            
                                    # 修改：稳定步数累加，达到阈值后才增大 time_step
                                    stable_steps += 1
                                    print(f'Stable steps after convergence: {stable_steps}/{STABLE_STEPS_REQUIRED}')
                                    if stable_steps >= STABLE_STEPS_REQUIRED:
                                            time_step = 1.5 * time_step
                                            if time_step > max_time_step:
                                                    time_step = max_time_step
                                            stable_steps = 0  # 重置计数器
                                            print(f'New time step after {STABLE_STEPS_REQUIRED} stable steps: {time_step:.6e} s')
                                    else:
                                            print(f'Holding time step at: {time_step:.6e} s (waiting for {STABLE_STEPS_REQUIRED} stable steps)')

                                    if time_now == t_end:
                                            break
                                    elif time_now + time_step <= t_end:
                                            time_now += time_step
                                    elif time_now + time_step > t_end:
                                            time_now = t_end
                            else:
                                    print(f'Simulation did not converge at time: {time_now:.4f} s and voltage: {vapplied:.4f} V')
                                    simulator = copy.deepcopy(simulator_last)

                                    time_step = time_step/1.5
                                    print(f'New time step after non-convergence: {time_step:.6e} s')

                                    time_now = time_last + time_step
                            

with open(file_name_time_stepped_pvd, "a") as f:
        f.write("  </Collection>\n")
        f.write("</VTKFile>\n")

simulator_set=copy.deepcopy(simulator)

trace_set=go.Scatter(x=voltage, y=np.abs(current), mode='markers+lines', 
                line=dict(color=colors[2]), name='RRAM_set')

trace_v_time=go.Scatter(x=time_all, y=voltage, mode='markers+lines', 
                line=dict(color=colors[0]), name='Vapplied')
trace_vte_time=go.Scatter(x=time_all, y=voltage_vte, mode='markers+lines', 
                line=dict(color=colors[1]), name='Voltage on Device')

################################################################################################
#########################################  Serilization  #######################################
################################################################################################

file_name_plot = f"simulator_set.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(simulator_set, f)

file_name_plot = f"trace_set.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(trace_set, f)

file_name_plot = f"trace_v_time.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(trace_v_time, f)

file_name_plot = f"trace_vte_time.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(trace_vte_time, f)

file_name_plot = f"voltage.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(voltage, f)

file_name_plot = f"current.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(current, f)

file_name_plot = f"time_all.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(time_all, f)

file_name_plot = f"voltage_vte.pkl"
full_path_plot = os.path.join(vtk_foldername_states, file_name_plot)
with open(full_path_plot, 'wb') as f:
        dill.dump(voltage_vte, f)
```
